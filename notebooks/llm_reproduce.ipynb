# ACRA — reproduce the LLM tables from the raw log (no GPU)

```
import json,math
from collections import defaultdict
rows=[json.loads(l) for l in open("data/llm_four_model_raw.jsonl")]
def wil(k,n,z=1.96):
    if n==0:return(0,0,0)
    p=k/n;d=1+z*z/n;c=(p+z*z/(2*n))/d;h=(z*math.sqrt(p*(1-p)/n+z*z/(4*n*n)))/d;return(round(p,3),round(max(0,c-h),3),round(min(1,c+h),3))
```

## Table III — success by model x condition

```
agg=defaultdict(lambda:defaultdict(lambda:[0,0]))
for r in rows: a=agg[r["model"]][r["condition"]];a[1]+=1;a[0]+=int(r["attack_success"])
for m in ["qwen","mistral","phi","llama"]:
    print(m)
    for c in ["Clean","TEL","SEM","JOINT","DUAL-RAND"]:
        s,n=agg[m][c];print(f"  {c:10} {wil(s,n)}")
```

## Table IV — JOINT success by wording

```
jw=defaultdict(lambda:defaultdict(lambda:[0,0]))
for r in rows:
    if r["condition"]=="JOINT": a=jw[r["model"]][r["doc"]];a[1]+=1;a[0]+=int(r["attack_success"])
for m in ["qwen","mistral","phi","llama"]:
    for w in ["strong","medium","weak"]:
        s,n=jw[m][w]
        if n: print(f"{m:8} {w:7} {wil(s,n)}")
```
